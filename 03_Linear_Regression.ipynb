```python
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
```

```python
# 1. 인공 데이터셋 생성
true_w = torch.tensor([[2.0], [-3.4]])
true_b = 4.2
X = torch.randn(1000, 2)
y = X @ true_w + true_b + torch.randn(1000, 1) * 0.01
```

```python
# 2. 데이터 로더 설정 (미니배치 처리 자동화)
dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=10, shuffle=True)
```

```python
# 3. 모델, 손실 함수, 최적화 알고리즘 정의
model = nn.Linear(2, 1)     # 입력 차원 2, 출력 차원 1인 완전 연결 계층 (자동으로 초기화)
criterion = nn.MSELoss()    # 평균 제곱 오차(MSE) 손실 함수
optimizer = torch.optim.SGD(model.parameters(), lr=0.03) # 확률적 경사 하강법
```

```python
# 4. 모델 훈련 (Training Loop)
for epoch in range(3):
    for features, labels in dataloader:
        optimizer.zero_grad()                		# 1) 그래디언트 초기화
        loss = criterion(model(features), labels) 	# 2) 순전파 및 손실 계산
        loss.backward()                      		# 3) 역전파 (그래디언트 계산)
        optimizer.step()                            # 4) 파라미터 업데이트

    print(f'Epoch {epoch + 1}, Loss: {loss.item():.4f}')
```

```python
# 결과 확인 (가중치와 편향의 오차)
print(f"w 오차: {true_w - model.weight.T}")
print(f"b 오차: {true_b - model.bias.item()}")
```
