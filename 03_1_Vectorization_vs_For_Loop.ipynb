```python
import numpy as np
import time

# 벡터의 차원 설정
d = 1000000

# 1. 파이썬 기본 리스트 초기화 (모든 성분이 1)
a_list = [1.0] * d
b_list = [1.0] * d

# 2. NumPy 배열 초기화 (모든 성분이 1)
a_np = np.ones(d)
b_np = np.ones(d)

# --- for-loop를 이용한 순차적 덧셈 및 곱셈 연산 ---
start_time = time.time()
c_add_list = [0.0] * d
c_mul_list = [0.0] * d
for i in range(d):
    c_add_list[i] = a_list[i] + b_list[i]
    c_mul_list[i] = a_list[i] * b_list[i]
for_loop_time = time.time() - start_time

# --- 벡터화(NumPy)를 이용한 병렬 덧셈 및 곱셈 연산 ---
start_time = time.time()
c_add_np = a_np + b_np
c_mul_np = a_np * b_np
vectorized_time = time.time() - start_time

print(f"for-loop 소요 시간: {for_loop_time:.6f} 초")
print(f"벡터화 소요 시간: {vectorized_time:.6f} 초")
print(f"성능 향상: 약 {for_loop_time / vectorized_time:.2f} 배 빠른 처리 속도")
```
